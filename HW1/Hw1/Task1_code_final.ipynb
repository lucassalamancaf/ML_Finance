### 1.2 Returns, EWMA Volatility, Pairwise Granger Tables, and Networks

```python
ALPHA = 0.05
MAX_LAG = 5
EWMA_LAMBDA = 0.94


def resample_price_panel(price_panel, freq):
    """Last available price in each weekly/monthly period, clipped to the assigned epoch."""
    sampled = price_panel.resample(freq).last().ffill().dropna(how="any")
    return sampled.loc[:pd.Timestamp(END_DATE)]


def log_returns(price_panel):
    return np.log(price_panel).diff().dropna(how="any")


def ewma_volatility(return_panel, lambda_=EWMA_LAMBDA):
    """EWMA volatility sqrt(EWMA(r^2)); lambda=0.94 follows the usual RiskMetrics/Lab setup."""
    return return_panel.pow(2).ewm(alpha=1 - lambda_, adjust=False).mean().pow(0.5).dropna(how="any")


def _select_aic_lag(pair_data, max_lag=MAX_LAG):
    """Select an AIC lag in [1, max_lag] for one asset pair using VAR.select_order."""
    pair_data = pair_data.dropna()
    if pair_data.shape[0] < 8 or pair_data.nunique().min() <= 1:
        return None

    feasible_max = min(max_lag, max(1, (pair_data.shape[0] - 1) // 4))
    for trial_max in range(feasible_max, 0, -1):
        try:
            selected = VAR(pair_data).select_order(maxlags=trial_max, trend="n").selected_orders.get("aic")
            if selected is None or pd.isna(selected):
                selected = 1
            return int(min(max(int(selected), 1), trial_max))
        except Exception:
            continue
    return None


def _granger_pvalue(pair_data, lag):
    """Statsmodels tests whether column 2 Granger-causes column 1."""
    try:
        result = grangercausalitytests(pair_data, maxlag=lag, verbose=False)
        return float(result[lag][0]["ssr_ftest"][1])
    except Exception:
        return np.nan


def build_causality_table(series_panel, assets=ASSETS, alpha=ALPHA, max_lag=MAX_LAG):
    """
    Return row-to-column lag and p-value matrices.
    Entry lag_table.loc[cause, effect] = selected lag if cause -> effect is significant, else 0.
    """
    lag_table = pd.DataFrame(0, index=assets, columns=assets, dtype=int)
    pvalue_table = pd.DataFrame(np.nan, index=assets, columns=assets, dtype=float)
    selected_lag_table = pd.DataFrame(np.nan, index=assets, columns=assets, dtype=float)

    for cause in assets:
        for effect in assets:
            if cause == effect:
                continue
            pair = series_panel[[effect, cause]].replace([np.inf, -np.inf], np.nan).dropna()
            lag = _select_aic_lag(pair, max_lag=max_lag)
            if lag is None:
                continue
            pvalue = _granger_pvalue(pair, lag)
            selected_lag_table.loc[cause, effect] = lag
            pvalue_table.loc[cause, effect] = pvalue
            if np.isfinite(pvalue) and pvalue < alpha:
                lag_table.loc[cause, effect] = lag

    return lag_table, pvalue_table, selected_lag_table


def display_lag_table(lag_table, title):
    table = lag_table.astype(object).copy()
    for asset in table.index:
        table.loc[asset, asset] = "-"
    print(f"\n{title}")
    print(tabulate(table, headers="keys", tablefmt="rounded_outline"))


def network_summary(lag_table):
    adjacency = lag_table.fillna(0).astype(int)
    graph = nx.from_pandas_adjacency(adjacency, create_using=nx.DiGraph)
    rows = []
    for node in adjacency.index:
        rows.append({
            "asset": node,
            "out_degree": graph.out_degree(node),
            "in_degree": graph.in_degree(node),
        })
    return pd.DataFrame(rows).sort_values(["out_degree", "in_degree"], ascending=[False, False])


def plot_causality_network(lag_table, title):
    """
    Plot a directed Granger causality network matching the reference style:
    - Circular nodes, size = out-degree (driving power)
    - Light periwinkle fill with dark border
    - Labels inside nodes
    - Directed arrows, thin grey edges
    - White background, no axes
    """
    adjacency = lag_table.fillna(0).astype(int)
    graph = nx.from_pandas_adjacency(adjacency, create_using=nx.DiGraph)
    graph.remove_edges_from(
        [(u, v) for u, v, d in graph.edges(data=True) if d.get("weight", 0) == 0]
    )

    fig, ax = plt.subplots(figsize=(7, 6), facecolor="white")
    ax.set_facecolor("white")

    if graph.number_of_edges() == 0:
        ax.text(0.5, 0.5, "No significant causal edges at 5%",
                ha="center", va="center", fontsize=12)
        ax.set_title(title, fontsize=10, pad=10)
        ax.set_axis_off()
        plt.tight_layout()
        plt.show()
        return graph

    # ── Layout: shell puts the most-connected nodes in the centre ──────────
    pos = nx.spring_layout(graph, seed=42, k=2.0)

    # ── Node sizing: base + driving-power bonus ────────────────────────────
    out_deg = dict(graph.out_degree())
    max_out = max(out_deg.values()) if max(out_deg.values()) > 0 else 1
    node_sizes = [
        max(600, 600 + 1800 * (out_deg.get(n, 0) / max_out))
        for n in graph.nodes
    ]

    # ── Draw nodes ────────────────────────────────────────────────────────
    nx.draw_networkx_nodes(
        graph, pos,
        node_size=node_sizes,
        node_color="#b0b8e8",      # light periwinkle/lavender fill
        edgecolors="#2a2a5a",      # dark navy border
        linewidths=1.5,
        ax=ax,
    )

    # ── Draw labels inside nodes ──────────────────────────────────────────
    nx.draw_networkx_labels(
        graph, pos,
        font_size=7.5,
        font_color="#1a1a3a",
        font_weight="bold",
        ax=ax,
    )

    # ── Draw edges ────────────────────────────────────────────────────────
    nx.draw_networkx_edges(
        graph, pos,
        arrows=True,
        arrowsize=14,
        arrowstyle="-|>",
        width=0.9,
        edge_color="#555555",
        connectionstyle="arc3,rad=0.10",
        ax=ax,
        node_size=node_sizes,      # keeps arrows from overlapping nodes
        min_source_margin=18,
        min_target_margin=18,
    )

    # ── Title: two-line style matching the reference ──────────────────────
    sampling = "Weekly" if "Weekly" in title else "Monthly"
    series   = "Returns" if "return" in title.lower() else "Volatility"
    ax.set_title(
        f"Causal Network: {sampling} {series}\n(Node size = Driving Power)",
        fontsize=9, loc="left", pad=8,
    )
    ax.set_axis_off()
    plt.tight_layout()
    plt.show()
    return graph


def strongest_abs_correlations(panel, n=12):
    corr = panel.corr()
    pairs = []
    for i, left in enumerate(corr.columns):
        for right in corr.columns[i + 1:]:
            pairs.append({"pair": f"{left} - {right}", "corr": corr.loc[left, right], "abs_corr": abs(corr.loc[left, right])})
    return pd.DataFrame(pairs).sort_values("abs_corr", ascending=False).head(n).drop(columns="abs_corr")
```

```python
weekly_prices = resample_price_panel(prices_clean, "W-FRI")
monthly_prices = resample_price_panel(prices_clean, "ME")

weekly_returns = log_returns(weekly_prices)
monthly_returns = log_returns(monthly_prices)
weekly_ewma_vol = ewma_volatility(weekly_returns)
monthly_ewma_vol = ewma_volatility(monthly_returns)

print("Panel sizes used for Granger tests:")
print(tabulate(pd.DataFrame({
    "series": ["weekly_returns", "monthly_returns", "weekly_ewma_vol", "monthly_ewma_vol"],
    "rows": [len(weekly_returns), len(monthly_returns), len(weekly_ewma_vol), len(monthly_ewma_vol)],
    "start": [weekly_returns.index.min().date(), monthly_returns.index.min().date(), weekly_ewma_vol.index.min().date(), monthly_ewma_vol.index.min().date()],
    "end": [weekly_returns.index.max().date(), monthly_returns.index.max().date(), weekly_ewma_vol.index.max().date(), monthly_ewma_vol.index.max().date()],
}), headers="keys", tablefmt="rounded_outline", showindex=False))

q1_panels = {
    "Weekly returns": weekly_returns,
    "Monthly returns": monthly_returns,
    "Weekly EWMA volatility": weekly_ewma_vol,
    "Monthly EWMA volatility": monthly_ewma_vol,
}

q1_results = {}
for label, panel in q1_panels.items():
    lag_table, pvalue_table, selected_lag_table = build_causality_table(panel)
    q1_results[label] = {
        "panel": panel,
        "lag_table": lag_table,
        "pvalue_table": pvalue_table,
        "selected_lag_table": selected_lag_table,
    }
    display_lag_table(lag_table, f"{label}: significant Granger causality lags (alpha={ALPHA})")
```

```python
# ═══════════════════════════════════════════════════════════════════
#  Stationary Bootstrap Validation of Granger-Causality Links
# ═══════════════════════════════════════════════════════════════════
# Addresses multiple-testing concerns: with 90 pairwise tests per network
# at α=5%, ~4–5 false positives are expected by chance. The stationary
# bootstrap (Politis & Romano 1994) resamples blocks of observations,
# preserving local serial dependence in returns/volatility.
#
# A link is "bootstrap-supported" if:
#   - bootstrap p-value < 0.10  (test statistic rarely as extreme under H0)
#   - robustness score >= 0.10  (significant in >=10% of bootstrap samples)

from arch.bootstrap import StationaryBootstrap

N_BOOT       = 200      # number of bootstrap replications
BOOT_PVAL    = 0.10     # bootstrap p-value threshold
BOOT_ROBUST  = 0.10     # robustness score threshold
RNG_SEED     = 42


def _granger_fstat(pair_data, lag):
    """Return the SSR F-statistic for column 2 -> column 1 at given lag."""
    try:
        res = grangercausalitytests(pair_data, maxlag=lag, verbose=False)
        return float(res[lag][0]["ssr_ftest"][0])
    except Exception:
        return np.nan


def bootstrap_link(panel, cause, effect, lag, n_boot=N_BOOT, seed=RNG_SEED):
    """Stationary bootstrap of a single Granger link at the AIC-selected lag."""
    pair = panel[[effect, cause]].replace([np.inf, -np.inf], np.nan).dropna()
    if pair.shape[0] < max(8, 4 * lag):
        return np.nan, np.nan, np.nan

    # Original F-statistic on the actual data
    f_orig = _granger_fstat(pair, lag)
    if not np.isfinite(f_orig):
        return np.nan, np.nan, np.nan

    # Average block length: rule of thumb ~ n^(1/3) for stationary bootstrap
    block_len = max(2, int(round(pair.shape[0] ** (1 / 3))))

    sb = StationaryBootstrap(block_len, pair.values, seed=seed)
    f_boot   = []
    sig_boot = []
    for sample in sb.bootstrap(n_boot):
        boot_data = pd.DataFrame(sample[0][0], columns=pair.columns)
        f_b = _granger_fstat(boot_data, lag)
        if np.isfinite(f_b):
            f_boot.append(f_b)
            # Re-test significance on bootstrap sample
            p_b = _granger_pvalue(boot_data, lag)
            sig_boot.append(int(np.isfinite(p_b) and p_b < ALPHA))

    if not f_boot:
        return f_orig, np.nan, np.nan

    boot_pval  = float(np.mean(np.array(f_boot) >= f_orig))
    robustness = float(np.mean(sig_boot))
    return f_orig, boot_pval, robustness


def validate_network_with_bootstrap(panel, lag_table, label):
    """Run the stationary bootstrap on every significant link in lag_table."""
    rows = []
    sig_links = [(c, e, int(lag_table.loc[c, e]))
                 for c in ASSETS for e in ASSETS
                 if c != e and lag_table.loc[c, e] > 0]

    if not sig_links:
        print(f"  {label}: no significant links to validate.")
        return pd.DataFrame()

    print(f"  {label}: bootstrapping {len(sig_links)} significant link(s)...")
    corr_mat = panel.corr()
    for cause, effect, lag in sig_links:
        f_orig, boot_p, robust = bootstrap_link(panel, cause, effect, lag)
        supported = (np.isfinite(boot_p) and np.isfinite(robust) and
                     boot_p < BOOT_PVAL and robust >= BOOT_ROBUST)
        rows.append({
            "network":      label,
            "link":         f"{cause} -> {effect}",
            "lag":          lag,
            "correlation":  round(float(corr_mat.loc[cause, effect]), 3),
            "robustness":   f"{robust*100:.1f}%" if np.isfinite(robust) else "n/a",
            "boot_pvalue":  round(boot_p, 3) if np.isfinite(boot_p) else np.nan,
            "supported":    "yes" if supported else "no",
        })
    return pd.DataFrame(rows)


# ── Run bootstrap on every network ─────────────────────────────────
np.random.seed(RNG_SEED)
print("Running stationary bootstrap validation (this takes ~1–3 minutes)...\n")

boot_results = {}
for label, result in q1_results.items():
    boot_results[label] = validate_network_with_bootstrap(
        result["panel"], result["lag_table"], label
    )

# ── Combined table ─────────────────────────────────────────────────
all_boot = pd.concat(boot_results.values(), ignore_index=True)
print("\n" + "═" * 75)
print("FULL BOOTSTRAP RESULTS — all significant Granger links")
print("═" * 75)
print(tabulate(all_boot, headers="keys", tablefmt="rounded_outline", showindex=False))

# ── Surviving links only ───────────────────────────────────────────
supported = all_boot[all_boot["supported"] == "yes"].drop(columns="supported")
print("\n" + "═" * 75)
print(f"BOOTSTRAP-SUPPORTED LINKS  (boot p<{BOOT_PVAL}, robustness>={BOOT_ROBUST*100:.0f}%)")
print("═" * 75)
if len(supported) == 0:
    print("  No links survive the bootstrap criterion.")
else:
    print(tabulate(supported, headers="keys", tablefmt="rounded_outline", showindex=False))

# ── Summary counts ─────────────────────────────────────────────────
summary_rows = []
for label, df in boot_results.items():
    if len(df) == 0:
        summary_rows.append({"network": label, "n_significant": 0, "n_supported": 0, "survival_rate": "n/a"})
    else:
        n_sig = len(df)
        n_sup = (df["supported"] == "yes").sum()
        summary_rows.append({
            "network": label,
            "n_significant": n_sig,
            "n_supported": n_sup,
            "survival_rate": f"{n_sup/n_sig*100:.1f}%",
        })
print("\nSurvival rates per network:")
print(tabulate(pd.DataFrame(summary_rows), headers="keys", tablefmt="rounded_outline", showindex=False))
```

```python

from statsmodels.tsa.stattools import adfuller

adf_series = {
    'Price level: Index': gw_model['Index'],
    'Monthly log return: SP': gw_model[TARGET_RETURN],
    'Direction indicator: 1[SP > 0]': gw_model[TARGET_DIRECTION],
}

adf_rows = []
for label, series in adf_series.items():
    s = series.dropna()
    stat, pval, usedlag, nobs, crit, *_ = adfuller(s, autolag='AIC')
    adf_rows.append({
        'Target candidate': label,
        'ADF statistic': stat,
        'p-value': pval,
        'used lag': usedlag,
        'nobs': nobs,
        '5% critical value': crit['5%'],
        'stationary at 5%?': 'yes' if pval < 0.05 else 'no',
        'Role in our analysis': (
            'Rejected as main target: non-stationary level' if label.startswith('Price') else
            'Main regression target' if label.startswith('Monthly') else
            'Secondary classification target'
        ),
    })

adf_table = pd.DataFrame(adf_rows)
display(adf_table.round({'ADF statistic': 4, 'p-value': 4, '5% critical value': 4}))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
plot_adf = adf_table.copy()
bar_colors = ['firebrick' if p >= 0.05 else 'darkgreen' for p in plot_adf['p-value']]
axes[0].bar(plot_adf['Target candidate'], plot_adf['p-value'], color=bar_colors, alpha=0.85)
axes[0].axhline(0.05, color='black', linestyle='--', lw=0.9, label='5% rejection threshold')
axes[0].set_title('ADF Test p-values by Target Candidate')
axes[0].set_ylabel('ADF p-value')
axes[0].tick_params(axis='x', rotation=20)
axes[0].legend()

scaled_price = gw_model['Index'] / gw_model['Index'].iloc[0]
scaled_return_cum = np.exp(gw_model[TARGET_RETURN].cumsum())
rolling_up_share = gw_model[TARGET_DIRECTION].rolling(60).mean()
axes[1].plot(gw_model['Date'], scaled_price, label='Price level, indexed to 1', color='steelblue')
axes[1].plot(gw_model['Date'], scaled_return_cum / scaled_return_cum.iloc[0], label='Cumulative log-return index', color='darkgreen', alpha=0.75)
axes[1].plot(gw_model['Date'], rolling_up_share, label='5-year rolling up-month share', color='purple', alpha=0.8)
axes[1].set_title('Price Trends vs Return/Direction Transformations')
axes[1].legend()
plt.tight_layout()
plt.show()

print(
    'Target comparison: the price level fails the ADF stationarity test, so it is a poor main forecasting target. '
    'Log returns are stationary and economically interpretable for regression; the direction indicator is bounded and useful as a secondary classification target, but it discards magnitude information.'
)
```

### 2.3 Supervised Panels, Benchmarks, and Feature Selection

```python
from sklearn.base import clone
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.gaussian_process import GaussianProcessClassifier, GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, Matern, RationalQuadratic, WhiteKernel
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    mean_absolute_error,
    mean_squared_error,
    precision_score,
    recall_score,
    roc_auc_score,
    r2_score,
)
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.neural_network import MLPClassifier, MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.compose import TransformedTargetRegressor

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


def make_supervised_panel(target='return', predictors=None, lags=None, train_frac=TRAIN_FRAC):
    """Create a one-step-ahead panel: row t uses only variables lagged before target t."""
    if predictors is None:
        predictors = ASSIGNED_FEATURES
    if lags is None:
        lags = LAGS

    target_col = TARGET_RETURN if target == 'return' else TARGET_DIRECTION
    panel = gw_model[['Date', target_col] + list(dict.fromkeys(predictors))].copy()
    panel = panel.rename(columns={target_col: 'TARGET'})

    for lag in lags:
        panel[f'TARGET_lag{lag}'] = panel['TARGET'].shift(lag)
    for var in predictors:
        for lag in lags:
            panel[f'{var}_lag{lag}'] = panel[var].shift(lag)

    panel = panel.drop(columns=list(predictors), errors='ignore').replace([np.inf, -np.inf], np.nan).dropna().reset_index(drop=True)
    feature_cols = [c for c in panel.columns if c not in ['Date', 'TARGET']]
    split_idx = int(train_frac * len(panel))

    train = panel.iloc[:split_idx].copy()
    test = panel.iloc[split_idx:].copy()

    y_train = train['TARGET'].astype(int) if target == 'direction' else train['TARGET'].astype(float)
    y_test = test['TARGET'].astype(int) if target == 'direction' else test['TARGET'].astype(float)

    return {
        'panel': panel,
        'train': train,
        'test': test,
        'X_train_all': train[feature_cols],
        'X_test_all': test[feature_cols],
        'y_train': y_train,
        'y_test': y_test,
        'feature_cols': feature_cols,
        'target': target,
    }


def regression_metrics(name, y_true, y_pred, train_mean):
    y_pred = np.asarray(y_pred).reshape(-1)
    mse = mean_squared_error(y_true, y_pred)
    mse_mean = mean_squared_error(y_true, np.repeat(train_mean, len(y_true)))
    return {
        'model': name,
        'MSE': mse,
        'RMSE': np.sqrt(mse),
        'MAE': mean_absolute_error(y_true, y_pred),
        'R2': r2_score(y_true, y_pred),
        'Outperform mean %': 100 * (1 - mse / mse_mean),
        'Sign accuracy %': 100 * accuracy_score((np.asarray(y_true) > 0).astype(int), (y_pred > 0).astype(int)),
    }


def classification_metrics(name, y_true, score, threshold=0.5):
    score = np.asarray(score).reshape(-1)
    y_pred = (score >= threshold).astype(int)
    row = {
        'model': name,
        'Accuracy %': 100 * accuracy_score(y_true, y_pred),
        'Balanced accuracy %': 100 * balanced_accuracy_score(y_true, y_pred),
        'Precision %': 100 * precision_score(y_true, y_pred, zero_division=0),
        'Recall %': 100 * recall_score(y_true, y_pred, zero_division=0),
        'F1 %': 100 * f1_score(y_true, y_pred, zero_division=0),
        'ROC-AUC %': 100 * roc_auc_score(y_true, score) if len(np.unique(score)) > 1 else np.nan,
    }
    return row, y_pred


def estimator_scores(estimator, X):
    if hasattr(estimator, 'predict_proba'):
        return estimator.predict_proba(X)[:, 1]
    return np.asarray(estimator.predict(X)).reshape(-1)


def cv_permutation_feature_selection(data, top_k=6, n_splits=5):
    """Training-only nonlinear feature screening using expanding-window permutation importance."""
    X_train = data['X_train_all']
    y_train = data['y_train']
    tscv = TimeSeriesSplit(n_splits=n_splits)
    fold_importances = []

    for train_idx, val_idx in tscv.split(X_train):
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        fs_model = GradientBoostingRegressor(
            n_estimators=100,
            max_depth=2,
            learning_rate=0.05,
            subsample=0.8,
            random_state=RANDOM_STATE,
        )
        fs_model.fit(X_tr, y_tr)
        perm = permutation_importance(
            fs_model,
            X_val,
            y_val,
            n_repeats=5,
            random_state=RANDOM_STATE,
            scoring='neg_mean_squared_error',
        )
        fold_importances.append(perm.importances_mean)

    mean_importance = np.mean(fold_importances, axis=0)
    importance = pd.DataFrame({
        'feature': X_train.columns,
        'importance': mean_importance,
    }).sort_values('importance', ascending=False).reset_index(drop=True)

    selected = importance.head(top_k)['feature'].tolist()
    positive = importance[importance['importance'] > 0]['feature'].tolist()
    robust = positive[:max(top_k, min(10, len(positive)))] if positive else selected
    return importance, selected, robust

return_data = make_supervised_panel(target='return', predictors=ASSIGNED_FEATURES)

print(f"Return train period: {return_data['train']['Date'].min().date()} -> {return_data['train']['Date'].max().date()} ({len(return_data['train'])} months)")
print(f"Return test period : {return_data['test']['Date'].min().date()} -> {return_data['test']['Date'].max().date()} ({len(return_data['test'])} months)")
print(f"Candidate return features: {len(return_data['feature_cols'])}")

mean_benchmark_pred = np.repeat(return_data['y_train'].mean(), len(return_data['y_test']))
mean_benchmark = regression_metrics('Historical mean return', return_data['y_test'], mean_benchmark_pred, return_data['y_train'].mean())
display(pd.DataFrame([mean_benchmark]).set_index('model').round(6))

feature_importance, selected_top6, selected_robust = cv_permutation_feature_selection(return_data, top_k=6, n_splits=5)
print('Selected parsimonious feature set:', selected_top6)
print('Selected robust feature set:', selected_robust)
display(feature_importance.head(20).round(8))

fig, ax = plt.subplots(figsize=(9, 6))
plot_imp = feature_importance.sort_values('importance').tail(15)
ax.barh(plot_imp['feature'], plot_imp['importance'])
ax.axvline(0, color='black', lw=0.8)
ax.set_title('Training-Only CV Permutation Importance for Return Forecasting')
ax.set_xlabel('Mean drop in validation neg-MSE when feature is shuffled')
plt.tight_layout()
plt.show()

feature_sets = {
    'All assigned lags': return_data['feature_cols'],
    'Selected top 6': selected_top6,
    'Selected robust': selected_robust,
}
```

### 2.3b Modeling and Hyperparameter Rationale

The model grids are intentionally modest. Monthly return forecasting has a low signal-to-noise ratio, so highly flexible specifications are more likely to fit noise than stable predictability. The grids below are designed to satisfy the assignment's tuning requirement while keeping the comparison defensible and leakage-safe.

```python
rationale_table = pd.DataFrame([
    {
        'Component': 'Lag grid',
        'Choice': '1, 2, 3, 6, 12 months',
        'Why this is defensible': 'Combines short-run dynamics with quarterly, half-year, and annual macro effects.',
        'Overfitting control': 'Small fixed grid, applied equally to target lags and assigned predictor lags.',
    },
    {
        'Component': 'Feature selection',
        'Choice': 'Training-only CV permutation importance',
        'Why this is defensible': 'Allows nonlinear screens while avoiding test-set feedback.',
        'Overfitting control': 'Expanding-window validation and parsimonious/robust feature sets.',
    },
    {
        'Component': 'MLPRegressor',
        'Choice': 'Small/deep-narrow networks, tanh/relu, high L2, scaled target',
        'Why this is defensible': 'Monthly returns are tiny in magnitude; target scaling helps optimization and L2 discourages noise fitting.',
        'Overfitting control': 'TimeSeriesSplit tuning and conservative layer sizes.',
    },
    {
        'Component': 'GaussianProcessRegressor',
        'Choice': 'RBF, Matern, RationalQuadratic, WhiteKernel/alpha noise control',
        'Why this is defensible': 'Kernels encode smooth nonlinear relationships; noise terms acknowledge large unpredictable return component.',
        'Overfitting control': 'Compare explicit white noise with alpha regularization; normalize_y=True.',
    },
    {
        'Component': 'LSTMRegressor',
        'Choice': 'Lookbacks 3 and 12, small units, dropout, L2, early stopping',
        'Why this is defensible': 'Tests whether a short or annual sequence of macro states helps returns.',
        'Overfitting control': 'Small architecture, chronological validation, dropout, and early stopping.',
    },
    {
        'Component': 'Benchmarks',
        'Choice': 'Historical mean for returns; majority and previous direction for classification',
        'Why this is defensible': 'Simple baselines are hard to beat in efficient monthly equity returns.',
        'Overfitting control': 'Models must improve out of sample, not only in training/CV.',
    },
])
display(rationale_table)
```

### 2.4 Neural Network Return Regression

```python
tscv = TimeSeriesSplit(n_splits=3)
mlp_results = []
mlp_predictions = {}

mlp_grid = {
    'regressor__mlp__hidden_layer_sizes': [(8,), (16,), (8, 4)],
    'regressor__mlp__activation': ['relu', 'tanh'],
    'regressor__mlp__alpha': [0.1, 1.0, 5.0],
}

for fs_name, cols in feature_sets.items():
    X_train = return_data['X_train_all'][cols]
    X_test = return_data['X_test_all'][cols]
    y_train = return_data['y_train']
    y_test = return_data['y_test']

    mlp_est = TransformedTargetRegressor(
        regressor=Pipeline([
            ('scaler', StandardScaler()),
            ('mlp', MLPRegressor(solver='lbfgs', max_iter=1500, random_state=RANDOM_STATE)),
        ]),
        transformer=StandardScaler(),
    )
    search = GridSearchCV(
        mlp_est,
        param_grid=mlp_grid,
        cv=tscv,
        scoring='neg_mean_squared_error',
        n_jobs=1,
        refit=True,
    )
    search.fit(X_train, y_train)
    pred = search.predict(X_test)
    row = regression_metrics(f'MLPRegressor - {fs_name}', y_test, pred, y_train.mean())
    row['Best params'] = search.best_params_
    row['Feature count'] = len(cols)
    mlp_results.append(row)
    mlp_predictions[fs_name] = pred

mlp_table = pd.DataFrame(mlp_results).sort_values('Outperform mean %', ascending=False)
display(mlp_table.drop(columns=['Best params']).round(6))
for row in mlp_results:
    print(row['model'], row['Best params'])
```

### 2.5 Gaussian Process Return Regression

```python
gp_results = []
gp_predictions = {}
gp_stds = {}

gp_grid = [
    {
        'gp__kernel': [RBF(length_scale=1.0) + WhiteKernel(noise_level=0.1),
                       RationalQuadratic(length_scale=1.0, alpha=1.0) + WhiteKernel(noise_level=0.1)],
        'gp__alpha': [1e-6],
    },
    {
        'gp__kernel': [RBF(length_scale=1.0), Matern(length_scale=1.0, nu=1.5)],
        'gp__alpha': [1e-3, 1e-2, 1e-1],
    },
]

gp_best_estimators = {}
for fs_name, cols in feature_sets.items():
    X_train = return_data['X_train_all'][cols]
    X_test = return_data['X_test_all'][cols]
    y_train = return_data['y_train']
    y_test = return_data['y_test']

    gp_pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('gp', GaussianProcessRegressor(normalize_y=True, optimizer=None, random_state=RANDOM_STATE)),
    ])
    search = GridSearchCV(
        gp_pipe,
        param_grid=gp_grid,
        cv=TimeSeriesSplit(n_splits=3),
        scoring='neg_mean_squared_error',
        n_jobs=1,
        refit=True,
    )
    search.fit(X_train, y_train)
    pred, std = search.best_estimator_.predict(X_test, return_std=True)
    row = regression_metrics(f'GPRegressor - {fs_name}', y_test, pred, y_train.mean())
    row['Best params'] = search.best_params_
    row['Feature count'] = len(cols)
    gp_results.append(row)
    gp_predictions[fs_name] = pred
    gp_stds[fs_name] = std
    gp_best_estimators[fs_name] = search.best_estimator_

gp_table = pd.DataFrame(gp_results).sort_values('Outperform mean %', ascending=False)
display(gp_table.drop(columns=['Best params']).round(6))
for row in gp_results:
    print(row['model'], row['Best params'])

best_gp_row = max(gp_results, key=lambda r: r['Outperform mean %'])
best_gp_name = best_gp_row['model'].replace('GPRegressor - ', '')
fig, ax = plt.subplots(figsize=(14, 5))
test_dates = return_data['test']['Date']
y_test = return_data['y_test']
y_pred = gp_predictions[best_gp_name]
y_std = gp_stds[best_gp_name]
ax.plot(test_dates, y_test, label='Actual returns', color='black', alpha=0.35)
ax.plot(test_dates, y_pred, label=f'Best GP forecast ({best_gp_name})', color='darkgreen')
ax.fill_between(test_dates, y_pred - 1.96 * y_std, y_pred + 1.96 * y_std, color='darkgreen', alpha=0.12, label='95% GP interval')
ax.axhline(return_data['y_train'].mean(), color='red', linestyle='--', label='Historical mean')
ax.set_title('Best Gaussian Process Return Forecast with Predictive Interval')
ax.legend()
plt.tight_layout()
plt.show()
```

### 2.6 LSTM Return Regression

```python
from tensorflow.keras.regularizers import l2


def make_lstm_regression_sequences(X, y, timesteps):
    X_values = np.asarray(X, dtype=np.float32)
    y_values = np.asarray(y, dtype=np.float32).reshape(-1)
    X_seq, y_seq = [], []
    for i in range(timesteps, len(X_values)):
        X_seq.append(X_values[i - timesteps:i])
        y_seq.append(y_values[i])
    return np.asarray(X_seq, dtype=np.float32), np.asarray(y_seq, dtype=np.float32)


def fit_lstm_regressor(X_train_df, y_train_ser, X_test_df, y_test_ser, timesteps, units, layers, dropout):
    x_scaler = StandardScaler()
    y_scaler = StandardScaler()
    X_train_scaled = x_scaler.fit_transform(X_train_df)
    X_test_scaled = x_scaler.transform(X_test_df)
    y_train_scaled = y_scaler.fit_transform(np.asarray(y_train_ser).reshape(-1, 1)).reshape(-1)

    X_seq, y_seq = make_lstm_regression_sequences(X_train_scaled, y_train_scaled, timesteps)
    X_test_seq, y_test_seq = make_lstm_regression_sequences(X_test_scaled, y_test_ser, timesteps)

    val_start = int(0.85 * len(X_seq))
    X_fit, y_fit = X_seq[:val_start], y_seq[:val_start]
    X_val, y_val = X_seq[val_start:], y_seq[val_start:]

    tf.keras.backend.clear_session()
    tf.random.set_seed(RANDOM_STATE)
    model = Sequential()
    if layers == 1:
        model.add(LSTM(units, input_shape=(timesteps, X_train_df.shape[1]),
                       dropout=dropout, recurrent_dropout=dropout,
                       kernel_regularizer=l2(1e-3), recurrent_regularizer=l2(1e-3)))
    else:
        model.add(LSTM(units, return_sequences=True, input_shape=(timesteps, X_train_df.shape[1]),
                       dropout=dropout, recurrent_dropout=dropout,
                       kernel_regularizer=l2(1e-3), recurrent_regularizer=l2(1e-3)))
        model.add(LSTM(units, dropout=dropout, recurrent_dropout=dropout,
                       kernel_regularizer=l2(1e-3), recurrent_regularizer=l2(1e-3)))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mean_squared_error')

    early_stop = EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True)
    history = model.fit(
        X_fit, y_fit,
        validation_data=(X_val, y_val),
        epochs=60,
        batch_size=16,
        shuffle=False,
        verbose=0,
        callbacks=[early_stop],
    )

    val_pred_scaled = model.predict(X_val, verbose=0).reshape(-1)
    val_mse = mean_squared_error(y_scaler.inverse_transform(y_val.reshape(-1, 1)).reshape(-1),
                                 y_scaler.inverse_transform(val_pred_scaled.reshape(-1, 1)).reshape(-1))
    test_pred_scaled = model.predict(X_test_seq, verbose=0).reshape(-1)
    test_pred = y_scaler.inverse_transform(test_pred_scaled.reshape(-1, 1)).reshape(-1)

    return {
        'model': model,
        'history': history,
        'val_mse': val_mse,
        'test_pred': test_pred,
        'y_test_seq': y_test_seq,
        'config': {'timesteps': timesteps, 'units': units, 'layers': layers, 'dropout': dropout},
    }

lstm_configs = [
    {'timesteps': 3, 'units': 16, 'layers': 1, 'dropout': 0.2},
    {'timesteps': 12, 'units': 16, 'layers': 1, 'dropout': 0.2},
    {'timesteps': 12, 'units': 32, 'layers': 1, 'dropout': 0.2},
]

lstm_results = []
lstm_predictions = {}
for fs_name in ['Selected top 6', 'Selected robust']:
    cols = feature_sets[fs_name]
    X_train = return_data['X_train_all'][cols]
    X_test = return_data['X_test_all'][cols]
    y_train = return_data['y_train']
    y_test = return_data['y_test']

    runs = [fit_lstm_regressor(X_train, y_train, X_test, y_test, **cfg) for cfg in lstm_configs]
    best_run = min(runs, key=lambda r: r['val_mse'])
    row = regression_metrics(f'LSTMRegressor - {fs_name}', best_run['y_test_seq'], best_run['test_pred'], y_train.mean())
    row['Best params'] = best_run['config']
    row['Feature count'] = len(cols)
    row['Validation MSE'] = best_run['val_mse']
    lstm_results.append(row)
    lstm_predictions[fs_name] = best_run

lstm_table = pd.DataFrame(lstm_results).sort_values('Outperform mean %', ascending=False)
display(lstm_table.drop(columns=['Best params']).round(6))
for row in lstm_results:
    print(row['model'], row['Best params'])
```

The assignment asks us to compare the forecasting models from Problem 2 against an ARMA$(p,q)$ + GARCH$(1,1)$ baseline, together with a normalization transformation of the target. We use the same monthly S&P 500 log-return target and the same chronological train/test split as in Problem 2, so the comparison is fair. The ARMA component forecasts the conditional mean of returns using an expanding-window one-step-ahead refit; the GARCH component models conditional volatility in the ARMA residuals.


### 3.1 Baseline Design and Transformations

```python
from arch import arch_model
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf
from sklearn.preprocessing import PowerTransformer

# Use the same supervised return split as Q2. This keeps the ARMA-GARCH baseline aligned with the ML test period.
y_train_raw = return_data['y_train'].copy()
y_test_raw = return_data['y_test'].copy()
train_dates = return_data['train']['Date'].copy()
test_dates = return_data['test']['Date'].copy()

print(f"ARMA-GARCH train period: {train_dates.min().date()} -> {train_dates.max().date()} ({len(y_train_raw)} months)")
print(f"ARMA-GARCH test period : {test_dates.min().date()} -> {test_dates.max().date()} ({len(y_test_raw)} months)")
print(f"Historical mean benchmark MSE: {mean_squared_error(y_test_raw, np.repeat(y_train_raw.mean(), len(y_test_raw))):.6f}")

class ReturnTransformer:
    def __init__(self, name):
        self.name = name
        self.model = None

    def fit(self, y):
        values = np.asarray(y).reshape(-1, 1)
        if self.name == 'Raw log return':
            self.model = None
        elif self.name == 'Z-score':
            self.model = StandardScaler().fit(values)
        elif self.name == 'Yeo-Johnson':
            self.model = PowerTransformer(method='yeo-johnson', standardize=True).fit(values)
        else:
            raise ValueError(self.name)
        return self

    def transform(self, y):
        values = np.asarray(y).reshape(-1, 1)
        if self.model is None:
            return np.asarray(y, dtype=float)
        return self.model.transform(values).reshape(-1)

    def inverse_transform(self, y_transformed):
        values = np.asarray(y_transformed).reshape(-1, 1)
        if self.model is None:
            return np.asarray(y_transformed, dtype=float)
        return self.model.inverse_transform(values).reshape(-1)

transformations = ['Raw log return', 'Z-score', 'Yeo-Johnson']

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, name in zip(axes, transformations):
    transformer = ReturnTransformer(name).fit(y_train_raw)
    ytr_t = transformer.transform(y_train_raw)
    sns.histplot(ytr_t, bins=45, kde=True, ax=ax, color='steelblue')
    ax.set_title(name)
    ax.axvline(np.mean(ytr_t), color='red', linestyle='--', lw=0.9)
plt.suptitle('Training Return Distribution Under Alternative Transformations', y=1.03)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
plot_acf(y_train_raw, lags=24, ax=axes[0], title='ACF of Monthly Returns')
plot_acf(y_train_raw ** 2, lags=24, ax=axes[1], title='ACF of Squared Monthly Returns')
plt.tight_layout()
plt.show()
```

### 3.2 ARMA Order Selection and ARMA-GARCH Forecasting

```python
def select_arma_order(y, max_p=5, max_q=5, max_total_lag=6):
    rows = []
    best = {'aic': np.inf, 'order': None, 'fit': None}
    y = pd.Series(y).astype(float)
    for p in range(max_p + 1):
        for q in range(max_q + 1):
            if p + q > max_total_lag:
                rows.append({'p': p, 'q': q, 'AIC': np.nan})
                continue
            try:
                fit = ARIMA(y, order=(p, 0, q), trend='c').fit()
                rows.append({'p': p, 'q': q, 'AIC': fit.aic})
                if fit.aic < best['aic']:
                    best = {'aic': fit.aic, 'order': (p, q), 'fit': fit}
            except Exception:
                rows.append({'p': p, 'q': q, 'AIC': np.nan})
    aic_table = pd.DataFrame(rows).pivot(index='p', columns='q', values='AIC')
    return best, aic_table


def arma_expanding_window_forecast(y_train_transformed, y_test_transformed, order):
    """One-step-ahead ARMA forecasts with expanding-window refits.

    At test month t, the model is fitted only on training data plus realised test
    returns up to t-1. This is slower than state updating but is a stronger
    classical forecasting benchmark.
    """
    forecasts = []
    y_train_series = pd.Series(y_train_transformed).astype(float).reset_index(drop=True)
    y_test_series = pd.Series(y_test_transformed).astype(float).reset_index(drop=True)
    for i in range(len(y_test_series)):
        hist = pd.concat([y_train_series, y_test_series.iloc[:i]], ignore_index=True)
        try:
            fit = ARIMA(hist, order=(order[0], 0, order[1]), trend='c').fit()
            fc = fit.forecast(steps=1)
            forecasts.append(float(np.asarray(fc)[0]))
        except Exception:
            forecasts.append(float(hist.mean()))
    return np.asarray(forecasts)


def fit_garch_on_residuals(residuals, horizon):
    # Scaling by 100 is standard for monthly returns and improves numerical stability.
    scaled_resid = 100 * pd.Series(residuals).dropna().astype(float)
    try:
        garch = arch_model(scaled_resid, mean='Zero', vol='GARCH', p=1, q=1, dist='skewt', rescale=False)
        garch_fit = garch.fit(disp='off', show_warning=False)
        variance = garch_fit.forecast(horizon=horizon, reindex=False).variance.iloc[-1].to_numpy()
        sigma = np.sqrt(variance) / 100
        return garch_fit, sigma
    except Exception:
        return None, np.repeat(np.std(residuals), horizon)

arma_garch_results = []
arma_garch_predictions = {}
arma_garch_volatility = {}
aic_tables = {}

aic_fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, transformation in zip(axes, transformations):
    transformer = ReturnTransformer(transformation).fit(y_train_raw)
    y_train_t = transformer.transform(y_train_raw)
    y_test_t = transformer.transform(y_test_raw)

    best, aic_table = select_arma_order(y_train_t, max_p=5, max_q=5, max_total_lag=6)
    p_best, q_best = best['order']
    aic_tables[transformation] = aic_table

    sns.heatmap(aic_table, annot=True, fmt='.1f', cmap='viridis_r', ax=ax)
    ax.set_title(f'{transformation}\nBest ARMA({p_best},{q_best})')
    ax.set_xlabel('q')
    ax.set_ylabel('p')

    arma_fit = best['fit']
    y_pred_t = np.asarray(arma_expanding_window_forecast(y_train_t, y_test_t, (p_best, q_best))).reshape(-1)
    y_pred = np.asarray(transformer.inverse_transform(y_pred_t)).reshape(-1)
    y_pred = y_pred[:len(y_test_raw)]

    garch_fit, sigma_t = fit_garch_on_residuals(arma_fit.resid, horizon=len(y_test_raw))
    sigma_t = np.asarray(sigma_t).reshape(-1)[:len(y_test_raw)]
    arma_garch_predictions[transformation] = y_pred
    arma_garch_volatility[transformation] = sigma_t

    row = regression_metrics(
        f'ARMA({p_best},{q_best}) + GARCH(1,1) - {transformation}',
        y_test_raw,
        y_pred,
        y_train_raw.mean(),
    )
    row['ARMA order'] = f'({p_best},{q_best})'
    row['Transformation'] = transformation
    row['GARCH dist'] = 'Skewed Student-t'
    arma_garch_results.append(row)

plt.suptitle('Expanded AIC Order Selection for ARMA Mean Equation (p,q <= 5, p+q <= 6)', y=1.05)
plt.tight_layout()
plt.show()

arma_garch_table = pd.DataFrame(arma_garch_results).sort_values('Outperform mean %', ascending=False)
display(arma_garch_table.round(6))
```

### 3.7 Regime-Dummy Robustness Check

As a small robustness check, we test whether adding pre-defined market-stress information helps the ARMA-GARCH mean forecast. To avoid look-ahead bias, the indicators below are lagged: the forecast for month `t` may use whether month `t-1` was in a high-volatility or named-crisis state, but it is not allowed to know that month `t` itself is a crisis month. This is intentionally treated as a robustness exercise, not as the main specification, because named crisis periods are sparse and can easily become an ex-post story if used too aggressively.


```python

def parse_arma_order(order_value):
    if isinstance(order_value, tuple):
        return order_value
    cleaned = str(order_value).replace('(', '').replace(')', '').replace(' ', '')
    p, q = cleaned.split(',')
    return int(p), int(q)


def armax_expanding_window_forecast(y_train_transformed, y_test_transformed, order, X_train=None, X_test=None):
    """One-step-ahead ARMAX forecasts with expanding-window refits and optional exogenous variables."""
    forecasts = []
    y_train_series = pd.Series(y_train_transformed).astype(float).reset_index(drop=True)
    y_test_series = pd.Series(y_test_transformed).astype(float).reset_index(drop=True)

    if X_train is not None:
        X_train_df = pd.DataFrame(X_train).astype(float).reset_index(drop=True)
        X_test_df = pd.DataFrame(X_test).astype(float).reset_index(drop=True)
    else:
        X_train_df = None
        X_test_df = None

    for i in range(len(y_test_series)):
        hist_y = pd.concat([y_train_series, y_test_series.iloc[:i]], ignore_index=True)
        if X_train_df is None:
            hist_x = None
            next_x = None
        else:
            hist_x = pd.concat([X_train_df, X_test_df.iloc[:i]], ignore_index=True)
            next_x = X_test_df.iloc[[i]]
        try:
            fit = ARIMA(hist_y, exog=hist_x, order=(order[0], 0, order[1]), trend='c').fit(
                method_kwargs={'maxiter': 75}
            )
            fc = fit.forecast(steps=1, exog=next_x)
            forecasts.append(float(np.asarray(fc)[0]))
        except Exception:
            forecasts.append(float(hist_y.mean()))
    return np.asarray(forecasts)


# Build lagged regime indicators from realised information only.
regime_panel = return_data['panel'][['Date', 'TARGET']].copy().sort_values('Date').reset_index(drop=True)
regime_panel['realized_vol_12m_lag1'] = regime_panel['TARGET'].rolling(12).std().shift(1)

train_date_index = pd.to_datetime(train_dates).reset_index(drop=True)
test_date_index = pd.to_datetime(test_dates).reset_index(drop=True)
train_mask = regime_panel['Date'].isin(train_date_index)
vol_threshold = regime_panel.loc[train_mask, 'realized_vol_12m_lag1'].quantile(0.80)
regime_panel['high_vol_lag1'] = (regime_panel['realized_vol_12m_lag1'] > vol_threshold).astype(int)

crisis_periods = {
    'Dot-com / early-2000s stress': ('2000-03-01', '2002-10-01'),
    'Global financial crisis': ('2007-12-01', '2009-06-01'),
    'COVID shock': ('2020-02-01', '2020-04-01'),
}
crisis_now = pd.Series(False, index=regime_panel.index)
for start, end in crisis_periods.values():
    crisis_now = crisis_now | regime_panel['Date'].between(pd.Timestamp(start), pd.Timestamp(end))
regime_panel['named_crisis_lag1'] = crisis_now.shift(1).fillna(False).astype(int)

regime_cols = ['high_vol_lag1', 'named_crisis_lag1']
regime_features = regime_panel.set_index('Date')[regime_cols]
X_regime_train = regime_features.reindex(train_date_index).fillna(0).reset_index(drop=True)
X_regime_test = regime_features.reindex(test_date_index).fillna(0).reset_index(drop=True)

regime_counts = pd.DataFrame({
    'Feature': regime_cols,
    'Train active months': X_regime_train.sum().astype(int).values,
    'Test active months': X_regime_test.sum().astype(int).values,
})
display(regime_counts)

# Use the best Q3 transformation/order. Fallback values allow this cell to be run standalone for output refreshes.
regime_transformation = globals().get('best_transformation', 'Yeo-Johnson')
if 'best_arma_row' in globals():
    regime_order = parse_arma_order(best_arma_row['ARMA order'])
else:
    regime_order = (2, 3)

regime_transformer = ReturnTransformer(regime_transformation).fit(y_train_raw)
y_train_regime_t = regime_transformer.transform(y_train_raw)
y_test_regime_t = regime_transformer.transform(y_test_raw)

# Baseline prediction from the main Q3 section if available; otherwise recompute only the best baseline.
if 'arma_garch_predictions' in globals() and regime_transformation in arma_garch_predictions:
    baseline_pred_for_regime = np.asarray(arma_garch_predictions[regime_transformation]).reshape(-1)
    baseline_row = arma_garch_table.loc[arma_garch_table['Transformation'].eq(regime_transformation)].iloc[0].to_dict()
else:
    baseline_pred_t = armax_expanding_window_forecast(
        y_train_regime_t,
        y_test_regime_t,
        regime_order,
        X_train=None,
        X_test=None,
    )
    baseline_pred_for_regime = regime_transformer.inverse_transform(baseline_pred_t)
    baseline_row = regression_metrics(
        f'ARMA({regime_order[0]},{regime_order[1]}) + GARCH(1,1) - {regime_transformation}',
        y_test_raw,
        baseline_pred_for_regime,
        y_train_raw.mean(),
    )
    baseline_row['ARMA order'] = f'({regime_order[0]},{regime_order[1]})'
    baseline_row['Transformation'] = regime_transformation

regime_pred_t = armax_expanding_window_forecast(
    y_train_regime_t,
    y_test_regime_t,
    regime_order,
    X_train=X_regime_train,
    X_test=X_regime_test,
)
regime_pred = regime_transformer.inverse_transform(regime_pred_t)

try:
    armax_train_fit = ARIMA(
        y_train_regime_t,
        exog=X_regime_train,
        order=(regime_order[0], 0, regime_order[1]),
        trend='c',
    ).fit(method_kwargs={'maxiter': 100})
    _, regime_sigma = fit_garch_on_residuals(armax_train_fit.resid, horizon=len(y_test_raw))
except Exception:
    regime_sigma = np.repeat(np.std(y_train_raw), len(y_test_raw))

regime_row = regression_metrics(
    f'ARMAX({regime_order[0]},{regime_order[1]}) + GARCH(1,1) - {regime_transformation} + lagged regimes',
    y_test_raw,
    regime_pred,
    y_train_raw.mean(),
)
regime_row['ARMA order'] = f'({regime_order[0]},{regime_order[1]})'
regime_row['Transformation'] = regime_transformation
regime_row['Regime variables'] = ', '.join(regime_cols)

regime_robustness_table = pd.DataFrame([
    {**baseline_row, 'Specification': 'Baseline ARMA-GARCH'},
    {**regime_row, 'Specification': 'ARMAX-GARCH + lagged regimes'},
])
cols_to_show = ['Specification', 'MSE', 'RMSE', 'MAE', 'R2', 'Outperform mean %', 'Sign accuracy %', 'ARMA order', 'Transformation']
display(regime_robustness_table[cols_to_show].round(6))

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
plot_regime = regime_robustness_table.copy()
axes[0].bar(plot_regime['Specification'], plot_regime['MSE'], color=['purple', 'darkgreen'], alpha=0.8)
axes[0].axhline(mean_squared_error(y_test_raw, np.repeat(y_train_raw.mean(), len(y_test_raw))), color='red', linestyle='--', label='Historical mean MSE')
axes[0].set_title('Regime Robustness: Test MSE')
axes[0].set_ylabel('MSE')
axes[0].tick_params(axis='x', rotation=20)
axes[0].legend()

axes[1].plot(test_dates, y_test_raw, color='black', alpha=0.30, label='Actual return')
axes[1].plot(test_dates, baseline_pred_for_regime, color='purple', lw=1.2, label='Baseline ARMA-GARCH')
axes[1].plot(test_dates, regime_pred, color='darkgreen', lw=1.2, label='ARMAX-GARCH + regimes')
for label, (start, end) in crisis_periods.items():
    start_ts, end_ts = pd.Timestamp(start), pd.Timestamp(end)
    if end_ts >= test_dates.min() and start_ts <= test_dates.max():
        axes[1].axvspan(max(start_ts, test_dates.min()), min(end_ts, test_dates.max()), color='gray', alpha=0.12)
axes[1].set_title('Return Forecasts With Lagged Regime Indicators')
axes[1].legend()
plt.tight_layout()
plt.show()

mse_delta = regime_row['MSE'] - baseline_row['MSE']
print(f"Regime robustness MSE change vs baseline: {mse_delta:+.6f}")
if mse_delta < 0:
    print('The lagged regime variables improve test MSE in this robustness check, but the gain should be interpreted cautiously because crisis states are sparse.')
else:
    print('The lagged regime variables do not improve test MSE, which supports the main conclusion that return mean predictability is weak.')
```

**Regime robustness interpretation.** The lagged regime extension improves the Yeo-Johnson ARMA-GARCH test MSE from `0.001725` to `0.001705`, a small `0.34%` gain versus the historical-mean benchmark, with the same `62.83%` sign accuracy. This is a useful robustness result, but it should not replace the main specification: the crisis dummy is sparse, the gain is modest, and the result depends on pre-defined event windows. The safest interpretation is that lagged stress regimes contain some information about the conditional mean, while the broader conclusion remains that monthly S&P 500 return predictability is weak and easy to overfit.
